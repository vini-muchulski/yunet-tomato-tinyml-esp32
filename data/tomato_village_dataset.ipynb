```python
import os
from PIL import Image, ImageFile
from tqdm import tqdm

ImageFile.LOAD_TRUNCATED_IMAGES = True

DATASET_DIR = "/home/jovyan/work/Datasets/tomato_village/tomato_village"

VALID_EXTENSIONS = {".jpg", ".jpeg", ".png", ".bmp", ".gif", ".webp"}

all_files = [
    os.path.join(dirpath, fname)
    for dirpath, _, filenames in os.walk(DATASET_DIR)
    for fname in filenames
    if os.path.splitext(fname)[1].lower() in VALID_EXTENSIONS
]

fixed = failed = skipped = 0
```

```python
for fpath in tqdm(all_files, desc="Processando imagens"):
    try:
        with Image.open(fpath) as img:
            rgb = img.convert("RGB")
            rgb.info.pop("icc_profile", None)
            rgb.save(fpath)
            fixed += 1
    except Exception as e:
        print(f"FALHOU: {fpath} — {e}")
        failed += 1

print(f"\nProcessadas: {fixed} | Falhas: {failed} | Ignoradas: {skipped}")
```
